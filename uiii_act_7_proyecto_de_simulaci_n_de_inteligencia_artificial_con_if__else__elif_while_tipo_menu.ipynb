```python
# 20/mayo/2026

# ===================================
# PROGRAMADOR Y PROYECTO IA
# ===================================

print("===================================")
print(" PROGRAMADOR Y PROYECTO IA ")
print("===================================")

nombre = input("Ingrese el nombre del programador: ")
proyecto = input("Ingrese el nombre del proyecto: ")

while True:

    print("\n=========== MENU ===========")
    print("1. Revisar lenguaje del programador")
    print("2. Revisar experiencia del programador")
    print("3. Analizar proyecto")
    print("4. Salir")

    opcion = int(input("Seleccione una opcion: "))

    # -----------------------------------
    # OPCION 1
    # -----------------------------------

    if opcion == 1:

        lenguaje = input("Ingrese lenguaje de programacion: ")
        dificultad = input("Ingrese dificultad del proyecto (bajo/medio/alto): ")

        print("\nANALIZANDO LENGUAJE...\n")

        if lenguaje == "python" and dificultad == "alto":
            print("🤖 Programador apto para Inteligencia Artificial.")
            print("💻 Recomendacion IA: Machine Learning")

        elif lenguaje == "javascript":
            print("🌐 Programador especializado en paginas web.")
            print("💻 Recomendacion IA: Desarrollo Frontend")

        elif lenguaje == "java":
            print("☕ Programador apto para aplicaciones.")
            print("💻 Recomendacion IA: Sistemas empresariales")

        else:
            print("💻 Lenguaje valido para programacion general.")

    # -----------------------------------
    # OPCION 2
    # -----------------------------------

    elif opcion == 2:

        experiencia = int(input("Ingrese años de experiencia: "))
        equipo = input("¿Trabaja en equipo? (si/no): ")

        print("\nREVISANDO EXPERIENCIA...\n")

        if experiencia >= 5 and equipo == "si":
            print("🔥 Programador avanzado.")
            print("✅ Apto para liderar proyectos.")

        elif experiencia >= 2:
            print("📚 Programador intermedio.")
            print("💻 Puede trabajar en proyectos generales.")

        else:
            print("⚠️ Programador principiante.")
            print("📖 Necesita seguir practicando.")

    # -----------------------------------
    # OPCION 3
    # -----------------------------------

    elif opcion == 3:

        estado = input("Ingrese estado del proyecto: ")
        entrega = input("¿Proyecto entregado? (si/no): ")

        print("\nANALIZANDO PROYECTO...\n")

        if estado == "terminado" and entrega == "si":
            print("✅ Proyecto finalizado correctamente.")
            print("🤖 IA detecta proyecto exitoso.")

        elif estado == "proceso":
            print("🛠️ Proyecto en desarrollo.")
            print("💻 Continuar trabajando.")

        else:
            print("❌ Proyecto con errores.")
            print("⚠️ Se recomienda corregir el sistema.")

    # -----------------------------------
    # OPCION 4
    # -----------------------------------

    elif opcion == 4:

        print("\n===================================")
        print(" REPORTE FINAL ")
        print("===================================")

        print("👨‍💻 Programador:", nombre)
        print("📁 Proyecto:", proyecto)
        print("📚 Programa realizado por Axel Guzman NL=27")

        print("\n✅ Gracias por utilizar PROGRAMADOR Y PROYECTO IA")

        break

    # -----------------------------------
    # OPCION INVALIDA
    # -----------------------------------

    else:

        print("❌ Opcion no valida")
```
